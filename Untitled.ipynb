CHAPTER 1:-- Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder,OneHotEncoder,StandardScaler,OrdinalEncoder
from sklearn.linear_model import LinearRegression,LogisticRegression
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings("ignore")
from sklearn.metrics import f1_score,confusion_matrix,accuracy_score,r2_score
```

CHAPTER 2 :-- Data Import

```python
price_df = pd.read_csv("E-Commerce Sales Analytics.csv")
price_df
```

CHAPTER 3 :-- Data Cleaning

```python
def reduce_to_smallnumbers(x):
    n = (x-10000)
    return n 

price_df["order_id"] = price_df["order_id"].apply(reduce_to_smallnumbers)
price_df
```

```python
price_df.describe()
```

#just observing the unique values to mark it down with suitable labels and encoders (transform strings to numbers or dividing them into categories)

```python
price_df["region"].unique()
```

```python
price_df["product_category"].unique()
```

```python
price_df["payment_method"].unique()
```

```python
encode = OrdinalEncoder(categories=[['South', 'East', 'West', 'North']])
price_df["region"] = encode.fit_transform(price_df[["region"]]).astype(int)
price_df
```

```python
labels= LabelEncoder()
price_df["payment_method"] = labels.fit_transform(price_df[["payment_method"]])
price_df
```

```python
encode = OrdinalEncoder(categories=[['Beauty', 'Clothing', 'Electronics', 'Home']])
price_df["product_category"] = encode.fit_transform(price_df[["product_category"]]).astype(int)
price_df
```

```python
price_df.drop("order_date",axis=1,inplace=True)
price_df
```

#dropping unnecessary column which doesn't contribute into prediction work

CHAPTER 5  :--- Model Testing

```python
x = price_df.drop(['revenue'],axis=1)
y = price_df['revenue']
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

#Using Model 1 (Linear Regression)

```python
model = LinearRegression()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
linear_r2 = r2_score(y_test,y_pred)
linear_r2
```

#Using Model 2 (Logistic Regression)

```python
model = LogisticRegression()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
r2 = r2_score(y_test,y_pred)
r2
```

#Using Model 3 (GaussianNB)

```python
model_nb = GaussianNB()
model_nb.fit(x_train,y_train)
y_pred = model_nb.predict(x_test)
r2 = r2_score(y_test,y_pred)
r2
```

**since classifier models are clearly not working  so we are going to use Regression models .... later we will discuss why ....

#Using Model 4(KNeighbor)

```python
model_knn = KNeighborsRegressor(n_neighbors=5)
model_knn.fit(x_train,y_train)
y_pred = model_knn.predict(x_test)
knn_r2 = r2_score(y_test,y_pred)
knn_r2
```

#Using Model 5(DecisionTree)

```python
model_dt = DecisionTreeRegressor()
model_dt.fit(x_train,y_train)
y_pred = model_dt.predict(x_test)
dt_r2 = r2_score(y_test,y_pred)
dt_r2
```

#Using Model 6 (RandomForest)

```python
model_random = RandomForestRegressor()
model_random.fit(x_train,y_train)
y_pred = model_random.predict(x_test)
random_r2 = r2_score(y_test,y_pred)
random_r2
```

#Using Model 7 (support vector)

```python

model_sv = SVR(kernel='rbf')
model_sv.fit(x_train,y_train)
y_pred = model_sv.predict(x_test)
sv_r2 = r2_score(y_test,y_pred)
sv_r2
```

**comapring the accuracy scores for better evaluation what model should we use and when to use.......

```python
scores = np.array([linear_r2,dt_r2,random_r2,sv_r2,knn_r2])
scores
```

```python
score_comp = np.array(["linear_r2","dt_r2","random_r2","sv_r2","knn_r2"])
score_comp
```

```python
sns.barplot(x=score_comp,y=scores)
```
